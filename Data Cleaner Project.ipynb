# 3. Simple Imputer(Category)

```python
df_random = df.copy()

cols = ["age", "gender", "region", "bmi",
        "blood_pressure", "cholesterol", "glucose"]

for col in cols:
    df_random[col + "_missing"] = df_random[col].isnull().astype(int)
    mask = df_random[col].isnull()
    values = df_random[col].dropna()
```

```python
    if mask.sum() > 0 and len(values) > 0:
        df_random.loc[mask, col] = np.random.choice(
            values, size=mask.sum(), replace=True
        )
```

```python
print("Missing Indicator + Random Sample Imputation")
display(df_random.head())

print("Remaining Missing Values:")
print(df_random[cols].isnull().sum())
```

# 2. Z-Score Method

```python
from scipy.stats import zscore
```

```python
z = df_outlier[["cholesterol", "glucose"]].apply(zscore)

df_zscore = df_outlier[(z.abs() <= 3).all(axis=1)]

print("After Z-score:", df_zscore.shape)
```

# 3. IQR Method

```python
Q1 = df_outlier["bmi"].quantile(0.25)
Q3 = df_outlier["bmi"].quantile(0.75)

IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
Upper = Q3 + 1.5 * IQR

df_IQR = df_outlier[
    (df_outlier["bmi"] >= lower) &
    (df_outlier["bmi"] <= Upper)
    ]
print(df_IQR)
```

# 4. Percentile Method

```python
df_percentile = df_outlier.copy()

for col in ["bmi", "cholesterol", "glucose"]:
    low = df_percentile[col].quantile(0.01)
    high = df_percentile[col].quantile(0.99)

    df_percentile[col] = df_percentile[col].clip(low, high)
    
df_percentile.head()
```

# 5. Winsorization

```python
from scipy.stats.mstats import winsorize
```

```python
df_winsor = df_outlier.copy()

for col in ["bmi", "cholesterol", "glucose"]:
    df_winsor[col] = winsorize(
        df_winsor[col], limits=[0.01, 0.01]
    )
```

```python
df_winsor.head()
```

```python
print("Original Shape:", df_outlier.shape)
print("Z-score Shape:", df_zscore.shape)
print("IQR Shape:", df_IQR.shape)
print("Percentile Shape:", df_percentile.shape)
print("Winsorization Shape:", df_winsor.shape)

print("\nOriginal Summary:")
display(df_outlier[["bmi", "cholesterol", "glucose"]].describe())

print("\nAfter Winsorization:")
display(df_winsor[["bmi", "cholesterol", "glucose"]].describe())
```

# Part - C Final Clean Dataset

```python
from scipy.stats.mstats import winsorize
```

```python
df_clean = df.copy()

num_cols = df_clean.select_dtypes(include="number").columns
df_clean[num_cols] = df_clean[num_cols].fillna(
    df_clean[num_cols].mean()
)
```

```python
cat_cols = df_clean.select_dtypes(include="object").columns

for col in cat_cols:
    if df_clean[col].isnull().any():
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].mode()[0]
        )
```

```python
for col in ["bmi", "cholesterol", "glucose"]:
    df_clean[col] = winsorize(
        df_clean[col], limits=[0.01, 0.01]
    )
    
print("Dataset Shape:", df_clean.shape)
print("Remaining Null Values:", df_clean.isnull().sum().sum())
display(df_clean.head())
```

```python
df_clean.to_csv("final_cleaned_dataset.csv", index=False)
```
